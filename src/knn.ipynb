```python
knn = KNNVectorSearch(X_train_customer.values, Y_train_customer.values)
```

```python
X_test_customer = pd.read_csv(os.path.join(training_data_path, "X_test_customer.csv"))
x_knn_column_name_strings = 'slot_A,slot_B,slot_C,slot_D,slot_E,slot_F,slot_G,slot_H,slot_I,slot_J,slot_K,slot type_ĐÔI,popcorn_0,gender_Nam,address_Cẩm Lệ,address_Huyện Hòa Vang,address_Hải Châu,address_III thành phố Đà Nẵng cũ,address_Liên Chiểu,address_Ngũ Hành Sơn,address_Thanh Khê,address_công,address_Điểm cuối,address_ủy quận Nhì Đà Nẵng,address_nan,job_blue collar,job_specialist,job_student,job_white collar,industry_computer,industry_construction,industry_economics,industry_education,industry_engineering,industry_finance,industry_government agent,industry_health service,industry_social service,age_group_0-18,age_group_19-25,age_group_26-35,age_group_36-50,age_group_51-65'
x_knn_column_names = x_knn_column_name_strings.split(',')
X_test_customer = X_test_customer[x_knn_column_names]
X_test_customer.head()
```

```python
Y_test_customer = pd.read_csv(os.path.join(training_data_path, "Y_test_customer.csv"))
Y_test_customer.head()
```

```python
import numpy as np

def cosine_similarity(y_true, y_pred):
    dot = np.sum(y_true * y_pred, axis=1)
    norm_true = np.linalg.norm(y_true, axis=1)
    norm_pred = np.linalg.norm(y_pred, axis=1)
    return np.mean(dot / (norm_true * norm_pred + 1e-10))

def kl_divergence(y_true, y_pred, eps=1e-10):
    y_true = np.clip(y_true, eps, 1)
    y_pred = np.clip(y_pred, eps, 1)
    return np.mean(np.sum(y_true * np.log(y_true / y_pred), axis=1))

def cross_entropy(y_true, y_pred, eps=1e-10):
    y_pred = np.clip(y_pred, eps, 1 - eps)
    return np.mean(-np.sum(y_true * np.log(y_pred), axis=1))

def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

def mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))

# Chuẩn hóa Y_test về phân phối xác suất
Y_test = Y_test_customer.values
Y_test = Y_test / (np.sum(Y_test, axis=1, keepdims=True) + 1e-8)

# Truy vấn KNN
k = 5
results = knn.batch_query(X_test_customer.values, k=k)

# Tính trung bình có trọng số
Y_pred = []
for neighbors in results:
    labels = np.array([y for y, sim in neighbors])
    sims = np.array([sim for y, sim in neighbors])
    weighted_avg = np.average(labels, axis=0, weights=sims)
    weighted_avg /= (np.sum(weighted_avg) + 1e-8)  # chuẩn hóa thành phân phối
    Y_pred.append(weighted_avg)

Y_pred = np.array(Y_pred)

# Tính các chỉ số
mse_val = mse(Y_test, Y_pred)
mae_val = mae(Y_test, Y_pred)
cos_val = cosine_similarity(Y_test, Y_pred)
kl_val = kl_divergence(Y_test, Y_pred)
ce_val = cross_entropy(Y_test, Y_pred)

print(f"🔢 MSE:            {mse_val:.6f}")
print(f"🔢 MAE:            {mae_val:.6f}")
print(f"🔢 Cosine Sim:     {cos_val:.6f}")
print(f"🔢 KL Divergence:  {kl_val:.6f}")
print(f"🔢 Cross Entropy:  {ce_val:.6f}")
```
